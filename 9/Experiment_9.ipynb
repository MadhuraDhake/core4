WHAT IS A BOX PLOT (VERY IMPORTANT):

A box plot represents 5 key statistical values:

Minimum
First Quartile (Q1) → 25%
Median (Q2) → 50%
Third Quartile (Q3) → 75%
Maximum

FORMULAS:

Median:

Middle value after sorting
If odd: middle element
If even: average of middle two

Q1 (25th percentile):

Value below which 25% data lies

Q3 (75th percentile):

Value below which 75% data lies

IQR (Interquartile Range):

Formula:
IQR = Q3 - Q1

Outliers:

Lower bound = Q1 - 1.5 * IQR
Upper bound = Q3 + 1.5 * IQR

INTERNAL WORKING OF BOXPLOT:

Data is grouped by 'sex'
For each group:
Ages are sorted
Quartiles are calculated
Median is calculated
Outliers are detected
Then box is drawn:
Box = Q1 to Q3
Line inside = Median
Whiskers = min & max (excluding outliers)

WHAT WE ARE DOING IN THIS BLOCK:

Helps identify:

EXPLANATION :

sns.boxplot(...)

Again creating a box plot
But now with an additional parameter

x='sex'

Groups data based on gender (male / female)

y='age'

Shows age distribution

data=dataset

Uses Titanic dataset

hue="survived"

hue is used to add another level of grouping
'survived' is a column in dataset
Values:
0 → did not survive
1 → survived

WHAT THIS DOES:

Instead of just 2 boxes (male, female)
Now each gender is split into:
survived
not survived

So total groups become:

Male survived
Male not survived
Female survived
Female not survived

INTERNAL WORKING:

First group by sex
Then within each group, split by survived
For each subgroup:
Calculate quartiles
Draw separate box

WHY WE USE HUE:

To compare multiple categories simultaneously
Adds more insight without making multiple plots

WHAT WE ARE DOING IN THIS BLOCK:

Comparing:
Age distribution
Gender
Survival status

So basically analyzing:

"Did age and gender affect survival?"
