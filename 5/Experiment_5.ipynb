```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
dataset = pd.read_csv(r"Social_Network_Ads.csv")
```

```python
dataset.head()
```

```python
dataset.info()
```

```python
dataset.isnull().sum()
```

```python
dataset.shape
```

```python
x = dataset.iloc[:, [2, 3]].values
y = dataset.iloc[:, 4].values
```

```python
print(x)
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.30, random_state = 0)
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
x_train = sc.fit_transform(x_train)
x_test = sc.transform(x_test)
```

```python
from sklearn.linear_model import LogisticRegression
classifier = LogisticRegression(random_state = 0)
classifier.fit(x_train, y_train)
```

```python
y_pred = classifier.predict(x_test)
```

```python
print(y_pred)
```

```python
print(y_test)
```

```python
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test, y_pred)
print(cm)
```

BLOCK 10
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
x_train = sc.fit_transform(x_train)
x_test = sc.transform(x_test)

EXPLANATION :

StandardScaler is used for feature scaling.

Formula used:
Z = (X - Mean) / Standard Deviation

Where:

X = original value
Mean = average of feature
Std = spread of values

sc = StandardScaler() creates scaler object.

fit_transform(x_train):

fit → calculates mean and std
transform → applies scaling

transform(x_test):

uses SAME mean and std from training data

Important:
We do not fit test data to avoid data leakage.

What we are doing:
Normalizing data so features are on same scale.

classifier.fit(x_train, y_train)

EXPLANATION :

LogisticRegression is a classification algorithm.

Formula:
P = 1 / (1 + e^(-z))

Where:
z = b0 + b1x1 + b2x2

This is called sigmoid function.

classifier = LogisticRegression() creates model.

fit(x_train, y_train):

Learns relationship between input and output
Finds best coefficients

Internally:

Uses optimization to minimize error
Adjusts weights

What we are doing:
Training the model.

BLOCK 18
from sklearn.metrics import precision_score, recall_score
precision_score(y_test, y_pred)

EXPLANATION :

Precision formula:
Precision = TP / (TP + FP)

It tells:
Out of predicted positives, how many are correct.

BLOCK 19
recall_score(y_test, y_pred)

EXPLANATION :

Recall formula:
Recall = TP / (TP + FN)

It tells:
Out of actual positives, how many were correctly predicted.

BLOCK 20
from sklearn.metrics import f1_score
f1_score(y_test, y_pred)

EXPLANATION :

F1 Score formula:
F1 = 2 × (Precision × Recall) / (Precision + Recall)

It balances precision and recall.
