```python
import pandas as pd
import numpy as np
from sklearn import linear_model
from sklearn.model_selection import train_test_split
```

```python
from sklearn.datasets import fetch_california_housing

data = fetch_california_housing()
print(data)
```

```python
data.feature_names
```

```python
data.target
```

```python
x = pd.DataFrame(data.data, columns = data.feature_names)
y = pd.DataFrame(data.target)
```

```python
x.head(10)
```

```python
y.head(10)
```

```python
reg = linear_model.LinearRegression()
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.20, random_state=42)
```

```python
reg.fit(x_train, y_train)
```

```python
print(reg.coef_)
```

```python
y_pred = reg.predict(x_test)
print(y_pred)
```

We are building a Machine Learning Linear Regression model.

Goal:
We want to predict house prices based on different features like income, rooms, population, etc.

Process:

Load dataset
Convert it into structured format
Split data into training & testing
Train model using training data
Learn relationship using formula:
y = m1x1 + m2x2 + ... + mnxn + c
Predict values on unseen data
Check output

from sklearn → sklearn means scikit-learn, which is a machine learning library.

import linear_model → This imports the module that contains algorithms like Linear Regression.

from sklearn.model_selection → This module contains functions to split data.

WHY WE NEED THIS BLOCK:

We need tools:

pandas → to handle data

BLOCK 8
reg = linear_model.LinearRegression()

EXPLANATION :

reg → Variable storing model.

linear_model → Module containing regression algorithms.

LinearRegression() → Class used to create regression model.

Internally:
This creates an object with:

coefficients (m values)
intercept (c value)

Formula used:
y = m1x1 + m2x2 + ... + mnxn + c

WHY WE USE THIS:

To create model that learns relationship between inputs and output.

BLOCK 9
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.20, random_state=42)

EXPLANATION :

train_test_split() → Function to split dataset.

Arguments:

x → input data
y → output data

test_size = 0.20 → 20% data used for testing
Remaining 80% → training

random_state = 42 → Fixes randomness so result is same every time.

Internally:

Shuffles data
Splits into 2 parts:
Training → model learns
Testing → model is evaluated

Outputs:
x_train → training inputs
x_test → testing inputs
y_train → training outputs
y_test → testing outputs

WHY WE USE THIS:

To avoid overfitting and test model on unseen data.

BLOCK 10
reg.fit(x_train, y_train)

EXPLANATION :

fit() → Function used to train the model.

Arguments:
x_train → input features
y_train → actual output

Internally what happens:

The model calculates coefficients using formula:

m = (XᵀX)⁻¹ XᵀY

Step-by-step:

Takes input features (X matrix)
Uses mathematical optimization
Minimizes error (difference between actual and predicted)

Error formula:
Error = (Actual - Predicted)²

Goal:
Minimize sum of squared errors.

Result:
Model learns:

coefficients (weights)
intercept

WHY WE USE THIS:

This is the learning step.

BLOCK 11
print(reg.coef_)

EXPLANATION :

reg → trained model

.coef_ → attribute storing coefficients (m values)

These represent:
How much each feature affects output.

Example:
If coefficient is high → strong impact

print() → displays values.

WHY WE USE THIS:

To understand model behavior.

BLOCK 12
y_pred = reg.predict(x_test)
print(y_pred)

EXPLANATION :

predict() → Function used to predict outputs.

Argument:
x_test → unseen data

Internally:
Uses formula:
y = m1x1 + m2x2 + ... + c

Step-by-step:

Takes each row from x_test
Multiplies with coefficients
Adds intercept
Produces predicted value

y_pred → predicted outputs

print(y_pred) → displays predictions.

WHY WE USE THIS:

To see how well model predicts new data.

FINAL UNDERSTANDING (VERY IMPORTANT)

We are:

Taking housing data
Learning relationship between features & price
Creating a mathematical model
Using formula to predict house prices

Core Formula Used:

y = m1x1 + m2x2 + ... + mnxn + c

Where:

y → predicted output
m → coefficients
x → input features
c → intercept
