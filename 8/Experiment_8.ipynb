sns.histplot(dataset['fare'], kde=True)

EXPLANATION :

sns.histplot → Function used to create a histogram.

Histogram formula concept:
It divides data into bins and counts frequency:
Frequency = number of values in each interval

dataset['fare'] →
dataset → DataFrame
['fare'] → selecting the fare column

kde=True →
kde stands for Kernel Density Estimation.

KDE formula (concept):
Smooth probability curve approximation of distribution.

Internally what happens:
Data → divided into bins → count frequency → draw bars
Then KDE → smooth curve fitted over histogram.

What we are doing:
Visualizing distribution of fare with both bars and smooth curve.

Comparing average age based on gender.

sns.boxplot(x='sex', y='age', data=dataset)

EXPLANATION :

sns.boxplot → Shows distribution using quartiles.

Formula concepts:

Q1 = 25th percentile
Q2 = Median (50th percentile)
Q3 = 75th percentile

IQR = Q3 - Q1

Internally:
Box → between Q1 and Q3
Line → median
Whiskers → min and max

What we are doing:
Showing spread of age for each gender.
