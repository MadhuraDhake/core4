WHAT ARE WE DOING IN THIS CODE (Overall Understanding)

In this code, we are performing Exploratory Data Analysis (EDA) on the Iris dataset.

EDA means:
We are trying to understand the dataset using statistics and visualizations before applying any machine learning.

We are:

Loading the dataset
Viewing the data
Understanding structure (rows, columns, data types)
Calculating statistical values (mean, min, max, etc.)
Visualizing data using histograms and boxplots

Goal:
To understand how different features like Sepal Length, Sepal Width, Petal Length, Petal Width are distributed and related.

BLOCK 4
data.describe()

EXPLANATION :

describe()
This function gives statistical summary of numerical columns.

Internally what happens:

For each numeric column, it calculates:

count → total number of values
mean → average

Formula for mean:
mean = (sum of all values) / (number of values)

std → standard deviation

Formula:
std = sqrt( (Σ(x - mean)²) / n )

min → smallest value
25% → first quartile
50% → median
75% → third quartile
max → largest value

Median formula:
If odd → middle value
If even → average of two middle values

Step-by-step internally:

Select numeric columns
For each column:
Calculate mean
Calculate standard deviation
Sort values to find quartiles
Display results in table

Why we use:
To understand distribution and spread of data.

BLOCK 10
sns.boxplot(data=data, x="SepalLengthCm")

EXPLANATION :

boxplot
Used to visualize spread of data and detect outliers.

Internally calculates:

Minimum
Q1 (25%)
Median (50%)
Q3 (75%)
Maximum

Formula:
IQR = Q3 - Q1

Outliers:
Values < Q1 - 1.5IQR
or > Q3 + 1.5IQR

Step-by-step:

Sort data
Find quartiles
Draw box from Q1 to Q3
Draw median line
Draw whiskers
Plot outliers as points

Purpose:
To detect skewness and outliers.
